### 🚀 Sharing Models with 🤗 Transformers & Hugging Face Hub

One of the most powerful features of the **🤗 Transformers** library is the ability to easily share models and tokenizers with the community.

Before you start, make sure you have an account at 👉 https://huggingface.co

---

#### 🔐 Login to Hugging Face

If you're working in a **notebook**, you can log in using:

```python
from huggingface_hub import notebook_login

notebook_login()
````

Or if you're in a **terminal**, run:

```bash
huggingface-cli login
```

---

#### ⬆️ Push your model to the Hub

Once you're logged in, you can upload your model using `push_to_hub()`:

```python
model.push_to_hub("my-awesome-model")
```

This will create a public repository under your username named:
`my-awesome-model`

---

#### ⬇️ Load your model from the Hub

Anyone can load your model with the following:

```python
from transformers import AutoModel

model = AutoModel.from_pretrained("your-username/my-awesome-model")
```

## Encoding text

`````python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("bert-base-cased")

encoded_input = tokenizer("Hello, I'm a single sentence!")
print(encoded_input)
`````

`````python
tokenizer.decode(encoded_input["input_ids"])
`````

`````python
encoded_input = tokenizer("How are you?", "I'm fine, thank you!")
print(encoded_input)
`````

`````python
encoded_input = tokenizer("How are you?", "I'm fine, thank you!", return_tensors="pt")
print(encoded_input)
`````

## Padding Inputs

`````python
encoded_input = tokenizer(
    ["How are you?", "I'm fine, thank you!"], padding=True, return_tensors="pt"
)
print(encoded_input)
`````

## Truncating Inputs

`````python
encoded_input = tokenizer(
    "This is a very very very very very very very very very very very very very very very very very very very very very very very very very very very very very very very very very very very very very very very very very very very very very very very very very long sentence.",
    truncation=True,
)
print(encoded_input["input_ids"])
`````

`````python
encoded_input = tokenizer(
    ["How are you?", "I'm fine, thank you!"],
    padding=True,
    truncation=True,
    max_length=5,
    return_tensors="pt",
)
print(encoded_input)
`````

## Adding special tokens

`````python
encoded_input = tokenizer("How are you?")
print(encoded_input["input_ids"])
tokenizer.decode(encoded_input["input_ids"])
`````
